```python
from dataclasses import dataclass

KB_PER_MB = 1024
MB_PER_GB = 1024
CADANGAN = 0.30  # cadangan 30%


@dataclass
class JenisData:
    nama: str
    jumlah_unit: int
    ukuran_kb: float  # rata-rata per unit, dalam KB

    @property
    def total_kb(self) -> float:
        return self.jumlah_unit * self.ukuran_kb

    @property
    def total_mb(self) -> float:
        return self.total_kb / KB_PER_MB


JUMLAH_UMKM = 50
PRODUK_PER_UMKM = 20
JUMLAH_PRODUK = JUMLAH_UMKM * PRODUK_PER_UMKM  # 1.000

DATA = [
    JenisData("Profil UMKM (teks)", JUMLAH_UMKM, 2),
    JenisData("Logo & foto bengkel", JUMLAH_UMKM * 3, 400),
    JenisData("Produk (teks)", JUMLAH_PRODUK, 3),
    JenisData("Foto produk", JUMLAH_PRODUK * 5, 500),
    JenisData("Video produk/proses", 150, 20 * KB_PER_MB),  # 20 MB
    JenisData("Kategori", 30, 0.5),
    JenisData("Pengguna", 2000, 1),
    JenisData("Pesanan/inquiry", 5000, 3),
    JenisData("Ulasan & rating", 3000, 1),
    JenisData("Artikel budaya (teks+foto)", 100, 810),
    JenisData("Log aktivitas", 100_000, 0.5),
]


def main() -> None:
    print(f"{'Jenis data':<30}{'Unit':>10}{'KB/unit':>12}{'Total (MB)':>14}")
    print("-" * 66)
    for d in DATA:
        print(f"{d.nama:<30}{d.jumlah_unit:>10,}{d.ukuran_kb:>12,.1f}{d.total_mb:>14,.2f}")

    total_mb = sum(d.total_mb for d in DATA)
    cadangan_mb = total_mb * CADANGAN
    grand_mb = total_mb + cadangan_mb

    print("-" * 66)
    print(f"{'Total kebutuhan':<52}{total_mb:>14,.2f} MB ({total_mb / MB_PER_GB:.2f} GB)")
    print(f"{'Cadangan 30%':<52}{cadangan_mb:>14,.2f} MB")
    print(f"{'Total + cadangan':<52}{grand_mb:>14,.2f} MB ({grand_mb / MB_PER_GB:.2f} GB)")


if __name__ == "__main__":
    main()
```

```python
import sqlite3

# Membuat koneksi ke database SQLite lokal di Colab
conn = sqlite3.connect('katalog_perak_celuk.db')
cursor = conn.cursor()

# Membuat tabel-tabel sesuai skema relasional katalog perak Celuk
# Catatan: SQLite tidak memiliki tipe ENUM secara native, sehingga kita ganti dengan CHECK constraint

# 1. Tabel Pengguna
cursor.execute('''
CREATE TABLE IF NOT EXISTS pengguna (
  id_pengguna   INTEGER PRIMARY KEY AUTOINCREMENT,
  nama          TEXT NOT NULL,
  email         TEXT NOT NULL UNIQUE,
  password_hash TEXT NOT NULL,
  peran         TEXT NOT NULL CHECK(peran IN ('admin','pengrajin','pembeli')) DEFAULT 'pembeli',
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP
);
''')

# 2. Tabel UMKM
cursor.execute('''
CREATE TABLE IF NOT EXISTS umkm (
  id_umkm       INTEGER PRIMARY KEY AUTOINCREMENT,
  id_pengguna   INTEGER NOT NULL,
  nama_usaha    TEXT NOT NULL,
  pemilik       TEXT NOT NULL,
  alamat        TEXT NOT NULL,
  no_whatsapp   TEXT,
  deskripsi     TEXT,
  path_logo     TEXT,
  latitude      REAL,
  longitude     REAL,
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
  FOREIGN KEY (id_pengguna) REFERENCES pengguna(id_pengguna)
);
''')

# 3. Tabel Kategori
cursor.execute('''
CREATE TABLE IF NOT EXISTS kategori (
  id_kategori   INTEGER PRIMARY KEY AUTOINCREMENT,
  nama_kategori TEXT NOT NULL UNIQUE
);
''')

# 4. Tabel Produk
cursor.execute('''
CREATE TABLE IF NOT EXISTS produk (
  id_produk     INTEGER PRIMARY KEY AUTOINCREMENT,
  id_umkm       INTEGER NOT NULL,
  id_kategori   INTEGER NOT NULL,
  nama_produk   TEXT NOT NULL,
  deskripsi     TEXT,
  harga         REAL NOT NULL CHECK (harga >= 0),
  berat_gram    REAL,
  kadar_perak   TEXT,                     -- contoh: 925
  stok          INTEGER NOT NULL DEFAULT 0,
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
  diubah_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
  FOREIGN KEY (id_umkm) REFERENCES umkm(id_umkm) ON DELETE CASCADE,
  FOREIGN KEY (id_kategori) REFERENCES kategori(id_kategori)
);
''')

# 5. Tabel Media Produk
cursor.execute('''
CREATE TABLE IF NOT EXISTS media_produk (
  id_media      INTEGER PRIMARY KEY AUTOINCREMENT,
  id_produk     INTEGER NOT NULL,
  jenis         TEXT NOT NULL CHECK(jenis IN ('foto','video')),
  path_file     TEXT NOT NULL,
  ukuran_kb     INTEGER,
  urutan        INTEGER NOT NULL DEFAULT 1,
  FOREIGN KEY (id_produk) REFERENCES produk(id_produk) ON DELETE CASCADE
);
''')

# 6. Tabel Pesanan
cursor.execute('''
CREATE TABLE IF NOT EXISTS pesanan (
  id_pesanan    INTEGER PRIMARY KEY AUTOINCREMENT,
  id_pengguna   INTEGER NOT NULL,
  status        TEXT NOT NULL CHECK(status IN ('inquiry','diproses','dikirim','selesai','batal')) DEFAULT 'inquiry',
  total_harga   REAL NOT NULL DEFAULT 0,
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
  FOREIGN KEY (id_pengguna) REFERENCES pengguna(id_pengguna)
);
''')

# 7. Tabel Detail Pesanan
cursor.execute('''
CREATE TABLE IF NOT EXISTS detail_pesanan (
  id_detail     INTEGER PRIMARY KEY AUTOINCREMENT,
  id_pesanan    INTEGER NOT NULL,
  id_produk     INTEGER NOT NULL,
  jumlah        INTEGER NOT NULL CHECK (jumlah > 0),
  harga_satuan  REAL NOT NULL,
  FOREIGN KEY (id_pesanan) REFERENCES pesanan(id_pesanan) ON DELETE CASCADE,
  FOREIGN KEY (id_produk) REFERENCES produk(id_produk)
);
''')

# 8. Tabel Ulasan
cursor.execute('''
CREATE TABLE IF NOT EXISTS ulasan (
  id_ulasan     INTEGER PRIMARY KEY AUTOINCREMENT,
  id_produk     INTEGER NOT NULL,
  id_pengguna   INTEGER NOT NULL,
  rating        INTEGER NOT NULL CHECK (rating BETWEEN 1 AND 5),
  komentar      TEXT,
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
  FOREIGN KEY (id_produk) REFERENCES produk(id_produk) ON DELETE CASCADE,
  FOREIGN KEY (id_pengguna) REFERENCES pengguna(id_pengguna)
);
''')

# 9. Tabel Artikel
cursor.execute('''
CREATE TABLE IF NOT EXISTS artikel (
  id_artikel    INTEGER PRIMARY KEY AUTOINCREMENT,
  judul         TEXT NOT NULL,
  isi           TEXT NOT NULL,
  path_gambar   TEXT,
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP
);
''')

# 10. Tabel Log Aktivitas
cursor.execute('''
CREATE TABLE IF NOT EXISTS log_aktivitas (
  id_log        INTEGER PRIMARY KEY AUTOINCREMENT,
  id_pengguna   INTEGER,
  aksi          TEXT NOT NULL,
  keterangan    TEXT,
  waktu         DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP
);
''')

# Membuat index pendukung
cursor.execute("CREATE INDEX IF NOT EXISTS idx_produk_umkm ON produk (id_umkm);")
cursor.execute("CREATE INDEX IF NOT EXISTS idx_produk_kategori ON produk (id_kategori);")
cursor.execute("CREATE INDEX IF NOT EXISTS idx_log_waktu ON log_aktivitas (waktu);")

conn.commit()

# Memvalidasi pembuatan tabel
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()

print("=== Database 'katalog_perak_celuk.db' Berhasil Dibuat ===")
print("Daftar tabel yang terbuat:")
for table in tables:
    print(f"- {table[0]}")

conn.close()
```

```python
from dataclasses import dataclass
import pandas as pd

KB_PER_MB = 1024
MB_PER_GB = 1024
CADANGAN = 0.30  # cadangan 30%

@dataclass
class JenisData:
    nama: str
    jumlah_unit: int
    ukuran_kb: float  # rata-rata per unit, dalam KB

    @property
    def total_kb(self) -> float:
        return self.jumlah_unit * self.ukuran_kb

    @property
    def total_mb(self) -> float:
        return self.total_kb / KB_PER_MB

JUMLAH_UMKM = 50
PRODUK_PER_UMKM = 20
JUMLAH_PRODUK = JUMLAH_UMKM * PRODUK_PER_UMKM  # 1.000

DATA = [
    JenisData("Profil UMKM (teks)", JUMLAH_UMKM, 2),
    JenisData("Logo & foto bengkel", JUMLAH_UMKM * 3, 400),
    JenisData("Produk (teks)", JUMLAH_PRODUK, 3),
    JenisData("Foto produk", JUMLAH_PRODUK * 5, 500),
    JenisData("Video produk/proses", 150, 20 * KB_PER_MB),  # 20 MB
    JenisData("Kategori", 30, 0.5),
    JenisData("Pengguna", 2000, 1),
    JenisData("Pesanan/inquiry", 5000, 3),
    JenisData("Ulasan & rating", 3000, 1),
    JenisData("Artikel budaya (teks+foto)", 100, 810),
    JenisData("Log aktivitas", 100_000, 0.5),
]

def main() -> None:
    # Membuat DataFrame untuk menyajikan data secara profesional
    df_data = pd.DataFrame({
        "Jenis Data": [d.nama for d in DATA],
        "Unit": [f"{d.jumlah_unit:,}" for d in DATA],
        "KB / Unit": [f"{d.ukuran_kb:,.1f}" for d in DATA],
        "Total (MB)": [d.total_mb for d in DATA]
    })

    total_mb = sum(d.total_mb for d in DATA)
    cadangan_mb = total_mb * CADANGAN
    grand_mb = total_mb + cadangan_mb

    # Menampilkan DataFrame utama
    print("="*70)
    print("ESTIMASI KONDISI AWAL (BULAN 0) - DETAILED")
    print("="*70)
    pd.options.display.float_format = "{:,.2f}".format
    display(df_data)

    # Menampilkan ringkasan kalkulasi
    df_ringkasan = pd.DataFrame({
        "Metrik Estimasi": [
            "Total Kebutuhan Bersih",
            "Alokasi Cadangan (30%)",
            "Grand Total Kebutuhan (Kapasitas Bersih + Cadangan)"
        ],
        "Kapasitas (MB)": [
            f"{total_mb:,.2f} MB",
            f"{cadangan_mb:,.2f} MB",
            f"{grand_mb:,.2f} MB"
        ],
        "Kapasitas (GB)": [
            f"{total_mb / MB_PER_GB:,.2f} GB",
            f"{cadangan_mb / MB_PER_GB:,.2f} GB",
            f"{grand_mb / MB_PER_GB:,.2f} GB"
        ]
    })

    print("\n" + "="*70)
    print("RINGKASAN TOTAL KAPASITAS")
    print("="*70)
    display(df_ringkasan)

if __name__ == "__main__":
    main()
```

```python
import sqlite3

# Membuat koneksi ke database SQLite lokal di Colab
conn = sqlite3.connect('katalog_perak_celuk.db')
cursor = conn.cursor()

# Membuat tabel-tabel sesuai skema relasional katalog perak Celuk
# Catatan: SQLite tidak memiliki tipe ENUM secara native, sehingga kita ganti dengan CHECK constraint

# 1. Tabel Pengguna
cursor.execute('''
CREATE TABLE IF NOT EXISTS pengguna (
  id_pengguna   INTEGER PRIMARY KEY AUTOINCREMENT,
  nama          TEXT NOT NULL,
  email         TEXT NOT NULL UNIQUE,
  password_hash TEXT NOT NULL,
  peran         TEXT NOT NULL CHECK(peran IN ('admin','pengrajin','pembeli')) DEFAULT 'pembeli',
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP
);
''')

# 2. Tabel UMKM
cursor.execute('''
CREATE TABLE IF NOT EXISTS umkm (
  id_umkm       INTEGER PRIMARY KEY AUTOINCREMENT,
  id_pengguna   INTEGER NOT NULL,
  nama_usaha    TEXT NOT NULL,
  pemilik       TEXT NOT NULL,
  alamat        TEXT NOT NULL,
  no_whatsapp   TEXT,
  deskripsi     TEXT,
  path_logo     TEXT,
  latitude      REAL,
  longitude     REAL,
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
  FOREIGN KEY (id_pengguna) REFERENCES pengguna(id_pengguna)
);
''')

# 3. Tabel Kategori
cursor.execute('''
CREATE TABLE IF NOT EXISTS kategori (
  id_kategori   INTEGER PRIMARY KEY AUTOINCREMENT,
  nama_kategori TEXT NOT NULL UNIQUE
);
''')

# 4. Tabel Produk
cursor.execute('''
CREATE TABLE IF NOT EXISTS produk (
  id_produk     INTEGER PRIMARY KEY AUTOINCREMENT,
  id_umkm       INTEGER NOT NULL,
  id_kategori   INTEGER NOT NULL,
  nama_produk   TEXT NOT NULL,
  deskripsi     TEXT,
  harga         REAL NOT NULL CHECK (harga >= 0),
  berat_gram    REAL,
  kadar_perak   TEXT,                     -- contoh: 925
  stok          INTEGER NOT NULL DEFAULT 0,
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
  diubah_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
  FOREIGN KEY (id_umkm) REFERENCES umkm(id_umkm) ON DELETE CASCADE,
  FOREIGN KEY (id_kategori) REFERENCES kategori(id_kategori)
);
''')

# 5. Tabel Media Produk
cursor.execute('''
CREATE TABLE IF NOT EXISTS media_produk (
  id_media      INTEGER PRIMARY KEY AUTOINCREMENT,
  id_produk     INTEGER NOT NULL,
  jenis         TEXT NOT NULL CHECK(jenis IN ('foto','video')),
  path_file     TEXT NOT NULL,
  ukuran_kb     INTEGER,
  urutan        INTEGER NOT NULL DEFAULT 1,
  FOREIGN KEY (id_produk) REFERENCES produk(id_produk) ON DELETE CASCADE
);
''')

# 6. Tabel Pesanan
cursor.execute('''
CREATE TABLE IF NOT EXISTS pesanan (
  id_pesanan    INTEGER PRIMARY KEY AUTOINCREMENT,
  id_pengguna   INTEGER NOT NULL,
  status        TEXT NOT NULL CHECK(status IN ('inquiry','diproses','dikirim','selesai','batal')) DEFAULT 'inquiry',
  total_harga   REAL NOT NULL DEFAULT 0,
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
  FOREIGN KEY (id_pengguna) REFERENCES pengguna(id_pengguna)
);
''')

# 7. Tabel Detail Pesanan
cursor.execute('''
CREATE TABLE IF NOT EXISTS detail_pesanan (
  id_detail     INTEGER PRIMARY KEY AUTOINCREMENT,
  id_pesanan    INTEGER NOT NULL,
  id_produk     INTEGER NOT NULL,
  jumlah        INTEGER NOT NULL CHECK (jumlah > 0),
  harga_satuan  REAL NOT NULL,
  FOREIGN KEY (id_pesanan) REFERENCES pesanan(id_pesanan) ON DELETE CASCADE,
  FOREIGN KEY (id_produk) REFERENCES produk(id_produk)
);
''')

# 8. Tabel Ulasan
cursor.execute('''
CREATE TABLE IF NOT EXISTS ulasan (
  id_ulasan     INTEGER PRIMARY KEY AUTOINCREMENT,
  id_produk     INTEGER NOT NULL,
  id_pengguna   INTEGER NOT NULL,
  rating        INTEGER NOT NULL CHECK (rating BETWEEN 1 AND 5),
  komentar      TEXT,
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP,
  FOREIGN KEY (id_produk) REFERENCES produk(id_produk) ON DELETE CASCADE,
  FOREIGN KEY (id_pengguna) REFERENCES pengguna(id_pengguna)
);
''')

# 9. Tabel Artikel
cursor.execute('''
CREATE TABLE IF NOT EXISTS artikel (
  id_artikel    INTEGER PRIMARY KEY AUTOINCREMENT,
  judul         TEXT NOT NULL,
  isi           TEXT NOT NULL,
  path_gambar   TEXT,
  dibuat_pada   DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP
);
''')

# 10. Tabel Log Aktivitas
cursor.execute('''
CREATE TABLE IF NOT EXISTS log_aktivitas (
  id_log        INTEGER PRIMARY KEY AUTOINCREMENT,
  id_pengguna   INTEGER,
  aksi          TEXT NOT NULL,
  keterangan    TEXT,
  waktu         DATETIME NOT NULL DEFAULT CURRENT_TIMESTAMP
);
''')

# Membuat index pendukung
cursor.execute("CREATE INDEX IF NOT EXISTS idx_produk_umkm ON produk (id_umkm);")
cursor.execute("CREATE INDEX IF NOT EXISTS idx_produk_kategori ON produk (id_kategori);")
cursor.execute("CREATE INDEX IF NOT EXISTS idx_log_waktu ON log_aktivitas (waktu);")

conn.commit()

# Memvalidasi pembuatan tabel
cursor.execute("SELECT name FROM sqlite_master WHERE type='table';")
tables = cursor.fetchall()

print("=== Database 'katalog_perak_celuk.db' Berhasil Dibuat ===")
print("Daftar tabel yang terbuat:")
for table in tables:
    print(f"- {table[0]}")

conn.close()
```

### **No. 2: Skenario Akuisisi Data & Estimasi Storage (1 Tahun)**

Topik yang dipilih adalah **Katalog UMKM Perak Desa Celuk**. Skenario akuisisi data dirancang berdasarkan estimasi pertumbuhan dan volume data berikut:

#### **Skenario Akuisisi Data**

1. **Pendaftaran UMKM & Produk**:
   * Pengrajin mendaftarkan profil UMKM mereka melalui formulir pendaftaran dan mengunggah produk secara berkala.
   * Diasumsikan terdapat **50 UMKM** terdaftar dengan masing-masing mengunggah **20 produk**.

   * Foto diunggah bersamaan dengan pengisian data produk.
   * Skenario alokasi media: **5 foto per produk** dan **3 foto bengkel per UMKM**.
   * Foto dikompresi otomatis ke format modern (WebP) dengan ukuran rata-rata **500 KB per file**.

3. **Manajemen Media Video**:
   * Bersifat opsional untuk mendemonstrasikan keaslian produk.
   * Durasi video dibatasi maksimal **30 detik** dalam format MP4 terkompresi dengan ukuran rata-rata **20 MB per file**.
   * Dialokasikan total **150 video** di dalam sistem.

4. **Interaksi Pembeli**:
   * Pembeli menghasilkan data transaksi berupa pesanan (*inquiry*) saat proses checkout, serta memberikan ulasan setelah barang diterima.

5. **Konten Edukasi & Budaya**:
   * Admin mengelola kategori produk serta menulis artikel budaya.
   * Target konten: **100 artikel**, di mana setiap artikel dilengkapi dengan teks penjelasan dan **2 foto pendukung**.

6. **Log Sistem**:
   * Sistem secara otomatis mencatat riwayat aktivitas pengguna melalui lapisan middleware keamanan.

7. **Metode Penyimpanan**:
   * **Database Relasional**: Digunakan untuk menyimpan seluruh data teks terstruktur.
   * **File Storage**: Digunakan untuk menyimpan seluruh aset biner (foto dan video). Database hanya menyimpan tautan (*path/URL*) menuju file terkait.

---

#### **Asumsi Pertumbuhan Data**
* Angka di atas merupakan kondisi awal (**Bulan 0**).
* Diasumsikan seluruh jenis data mengalami pertumbuhan kumulatif sebesar **10% per bulan** secara majemuk dengan formula:
  $$S(n) = S_0 \times 1.1^n$$
* Alokasi penyimpanan akhir ditambahkan **cadangan (buffer) sebesar 30%** untuk menjaga keandalan sistem.

```python
from dataclasses import dataclass
import os
import pandas as pd

# Pengaturan tampilan Pandas agar tabel rapi
pd.options.display.float_format = "{:,.2f}".format
pd.options.display.width = 250
pd.options.display.max_columns = 30

# ============================================================
# 1. KONSTANTA & PARAMETER ESTIMASI
# ============================================================
KB_PER_MB = 1024
MB_PER_GB = 1024
CADANGAN_PERSEN = 0.30       # Alokasi buffer cadangan 30%
PERTUMBUHAN_BULANAN = 0.10   # Tingkat pertumbuhan data 10% per bulan
DURASI_PROYEKSI = 12         # Proyeksi untuk jangka waktu 12 bulan

JUMLAH_UMKM = 50
PRODUK_PER_UMKM = 20
JUMLAH_PRODUK = JUMLAH_UMKM * PRODUK_PER_UMKM

# ============================================================
# 2. STRUKTUR DATA & ESTIMASI BULAN 0
# ============================================================
@dataclass
class KategoriData:
    nama: str
    jumlah_unit: int
    ukuran_kb_per_unit: float
    deskripsi: str = ""

    @property
    def total_kb(self) -> float:
        return self.jumlah_unit * self.ukuran_kb_per_unit

    @property
    def total_mb(self) -> float:
        return self.total_kb / KB_PER_MB


# Daftar rincian data kondisi awal
DAFTAR_DATA = [
    KategoriData("Profil UMKM (teks)",        JUMLAH_UMKM,         2.0,            "50 UMKM x 2 KB"),
    KategoriData("Logo & foto bengkel",       JUMLAH_UMKM * 3,     400.0,          "3 foto/UMKM x 400 KB"),
    KategoriData("Produk (teks)",             JUMLAH_PRODUK,       3.0,            "1.000 produk x 3 KB"),
    KategoriData("Foto produk",               JUMLAH_PRODUK * 5,   500.0,          "5 foto/produk x 500 KB (WebP)"),
    KategoriData("Video produk/proses",       150,                 20480.0,        "150 video x 20 MB (MP4, maks 30 detik)"),
    KategoriData("Kategori",                  30,                  0.5,            "30 kategori x 0,5 KB"),
    KategoriData("Pengguna",                  2000,                1.0,            "2.000 akun x 1 KB"),
    KategoriData("Pesanan/inquiry",           5000,                3.0,            "5.000 pesanan x 3 KB"),
    KategoriData("Ulasan & rating",           3000,                1.0,            "3.000 ulasan x 1 KB"),
    KategoriData("Artikel budaya (teks+foto)", 100,                810.0,          "100 artikel x 810 KB (teks + 2 foto)"),
    KategoriData("Log aktivitas",             100000,              0.5,            "100.000 baris x 0,5 KB"),
]

# Membuat DataFrame Kondisi Awal
tabel_awal = pd.DataFrame({
    "Jenis Data": [d.nama for d in DAFTAR_DATA],
    "Jumlah Unit": [d.jumlah_unit for d in DAFTAR_DATA],
    "Ukuran/Unit (KB)": [d.ukuran_kb_per_unit for d in DAFTAR_DATA],
    "Total (KB)": [d.total_kb for d in DAFTAR_DATA],
    "Total (MB)": [d.total_mb for d in DAFTAR_DATA],
    "Dasar Hitungan": [d.deskripsi for d in DAFTAR_DATA],
})
tabel_awal["Porsi (%)"] = (tabel_awal["Total (MB)"] / tabel_awal["Total (MB)"].sum()) * 100

total_awal_mb = tabel_awal["Total (MB)"].sum()
cadangan_awal_mb = total_awal_mb * CADANGAN_PERSEN

print("=== A. KONDISI AWAL (BULAN 0) ===")
display(tabel_awal)
print(f"Total Kebutuhan Bersih : {total_awal_mb:,.2f} MB ({total_awal_mb / MB_PER_GB:.2f} GB)")
print(f"Alokasi Cadangan (30%) : {cadangan_awal_mb:,.2f} MB")
print(f"Total + Cadangan       : {total_awal_mb + cadangan_awal_mb:,.2f} MB ({(total_awal_mb + cadangan_awal_mb) / MB_PER_GB:.2f} GB)\n")

# Validasi total awal dengan baseline
assert round(total_awal_mb, 2) == 5650.50, "Error: Total awal tidak sinkron dengan data kelompok!"

# ============================================================
# 3. PROYEKSI AKUMULASI STORAGE 12 BULAN
# ============================================================
list_proyeksi = []
kapasitas_sebelumnya = None

for bulan in range(DURASI_PROYEKSI + 1):
    faktor_pertumbuhan = (1 + PERTUMBUHAN_BULANAN) ** bulan
    total_mb = total_awal_mb * faktor_pertumbuhan
    tambahan_mb = 0.0 if kapasitas_sebelumnya is None else total_mb - kapasitas_sebelumnya
    total_dengan_cadangan_mb = total_mb * (1 + CADANGAN_PERSEN)

    list_proyeksi.append({
        "Bulan": bulan,
        "Faktor Pertumbuhan": faktor_pertumbuhan,
        "Total (MB)": total_mb,
        "Total (GB)": total_mb / MB_PER_GB,
        "Tambahan Bulanan (MB)": tambahan_mb,
        "Total + Cadangan (MB)": total_dengan_cadangan_mb,
        "Total + Cadangan (GB)": total_dengan_cadangan_mb / MB_PER_GB,
    })
    kapasitas_sebelumnya = total_mb

proyeksi_df = pd.DataFrame(list_proyeksi)
print("=== B. PROYEKSI TOTAL STORAGE 12 BULAN ===")
display(proyeksi_df)

# ============================================================
# 4. PROYEKSI DETAILED PER JENIS DATA (MB)
# ============================================================
proyeksi_per_jenis = pd.DataFrame(
    {
        f"Bulan {b}": [d.total_mb * (1 + PERTUMBUHAN_BULANAN) ** b for d in DAFTAR_DATA]
        for b in range(DURASI_PROYEKSI + 1)
    },
    index=[d.nama for d in DAFTAR_DATA],
)
proyeksi_per_jenis.loc["TOTAL"] = proyeksi_per_jenis.sum()

print("\n=== C. PROYEKSI PER JENIS DATA (MB) ===")
display(proyeksi_per_jenis)

# Validasi konsistensi kalkulasi
assert abs(proyeksi_per_jenis.loc["TOTAL", f"Bulan {DURASI_PROYEKSI}"] - proyeksi_df["Total (MB)"].iloc[-1]) < 1e-6

# ============================================================
# 5. RINGKASAN REKOMENDASI BULAN KE-12
# ============================================================
faktor_akhir = (1 + PERTUMBUHAN_BULANAN) ** DURASI_PROYEKSI
kapasitas_akhir_mb = total_awal_mb * faktor_akhir
kapasitas_akhir_cadangan_mb = kapasitas_akhir_mb * (1 + CADANGAN_PERSEN)
kapasitas_akhir_cadangan_gb = kapasitas_akhir_cadangan_mb / MB_PER_GB
tambahan_tahunan_mb = kapasitas_akhir_mb - total_awal_mb

ringkasan_df = pd.DataFrame({
    "Keterangan": [
        "Total Awal (Bulan 0)",
        "Faktor Pertumbuhan Komulatif 12 Bulan (1.1^12)",
        "Kebutuhan Bersih Bulan ke-12",
        "Akumulasi Penambahan Data 1 Tahun",
        "Alokasi Cadangan (30%) Bulan ke-12",
        "Kebutuhan Total Bulan ke-12 + Cadangan",
        "Rekomendasi Kapasitas Penyediaan (Pembulatan Atas)"
    ],
    "Nilai": [
        f"{total_awal_mb:,.2f} MB ({total_awal_mb / MB_PER_GB:.2f} GB)",
        f"{faktor_akhir:.4f} kali",
        f"{kapasitas_akhir_mb:,.2f} MB ({kapasitas_akhir_mb / MB_PER_GB:.2f} GB)",
        f"{tambahan_tahunan_mb:,.2f} MB ({tambahan_tahunan_mb / MB_PER_GB:.2f} GB)",
        f"{kapasitas_akhir_mb * CADANGAN_PERSEN:,.2f} MB",
        f"{kapasitas_akhir_cadangan_mb:,.2f} MB ({kapasitas_akhir_cadangan_gb:.2f} GB)",
        f"Minimal {int(kapasitas_akhir_cadangan_gb) + 1} GB"
    ]
})

print("\n=== D. RINGKASAN ESTIMASI BULAN KE-12 ===")
display(ringkasan_df)

# ============================================================
# 6. PENYIMPANAN DATASET HASIL
# ============================================================
os.makedirs("docs", exist_ok=True)
tabel_awal.to_csv("docs/storage_kondisi_awal.csv", index=False)
proyeksi_df.to_csv("docs/storage_proyeksi_12_bulan.csv", index=False)
proyeksi_per_jenis.to_csv("docs/storage_per_jenis_data.csv")
ringkasan_df.to_csv("docs/storage_ringkasan.csv", index=False)
print("\n[INFO] Seluruh file estimasi berhasil diperbarui dan disimpan di folder 'docs/'.")
```

### **Mengapa Multimedia Sebaiknya Tidak Disimpan sebagai BLOB di Database**

Berkas multimedia (seperti gambar, logo, dan video) sebaiknya disimpan di **File Storage** (misalnya Google Cloud Storage), sedangkan database cukup mencatat **path (jalur file) dan metadatanya**.

Berikut adalah beberapa alasan utama mengapa pendekatan ini jauh lebih baik:

1. **Ukuran Database Menjadi Terlalu Besar**
   * Pada data katalog UMKM ini, berkas media (foto, video, dan logo) mencakup **5.500 MB** dari total **5.650 MB** (sekitar **97%**), sedangkan data teks hanya sekitar **71 MB**.
   * Jika semua media disimpan sebagai BLOB, ukuran database akan membengkak hingga **79 kali lipat** dan berpotensi mendekati **17 GB** pada bulan ke-12.

2. **Proses Backup dan Restore yang Lambat**
   * Setiap kali melakukan backup, sistem harus menyalin seluruh file media yang berukuran besar, padahal file tersebut jarang sekali berubah. Hal ini membuat proses pemeliharaan database menjadi tidak efisien.

3. **Penurunan Performa Query**
   * File besar yang disimpan di dalam database akan memenuhi memori *cache* dan memperberat log transaksi saat proses unggah (*upload*).
   * Akibatnya, query sederhana (seperti menampilkan daftar produk) akan ikut melambat secara signifikan.

4. **Efisiensi Biaya (Cost & Scaling)**
   * Penyimpanan pada database jauh lebih mahal dan lebih sulit untuk ditingkatkan skalanya (*scaling*) dibandingkan dengan layanan *Object Storage* khusus seperti **Google Cloud Storage**.

5. **Kemudahan Penyajian ke Pengguna (Delivery & Speed)**
   * Berkas yang disimpan dalam bentuk BLOB harus dibaca melalui aplikasi backend terlebih dahulu sebelum dikirim ke pembeli, sehingga tidak bisa memanfaatkan fitur **CDN (Content Delivery Network)**, *browser caching*, atau *video streaming* secara langsung.

6. **Kemudahan Pengolahan File**
   * Proses manipulasi media seperti pembuatan gambar mini (*thumbnail*), kompresi otomatis, atau konversi format (misalnya ke WebP) jauh lebih mudah dilakukan jika file berada di penyimpanan biasa.

---

#### **Penerapan pada Skema Database Kita:**
* **Tabel `media_produk`**: Menyimpan referensi berupa `path_file`, `ukuran_kb`, dan `urutan` gambar.
* **Tabel `umkm` & `artikel`**: Kolom `path_logo` dan `path_gambar` hanya menyimpan string URL/path menuju file storage terkait.
* **Antisipasi Risiko**: Menyimpan file di luar database memiliki risiko munculnya *file yatim* (file tetap ada di storage meskipun datanya sudah terhapus di database). Hal ini dapat diatasi dengan mekanisme pembersihan terjadwal (*cron job*).
